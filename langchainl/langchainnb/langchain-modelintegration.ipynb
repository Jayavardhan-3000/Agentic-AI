```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0
)
response_groq = llm.invoke("What is Langchain?")
response_groq.content
```

**Streaming**

```python
llm.stream("Do rabbits love carrots?")
```

```python
for i in llm.stream("Do rabbits love carrots?"):
    print(i.text, end = "",flush = True)
```

**BATCH**

```python
llm.batch(
    [
        "Do rabbits exist on the moon?",
        "Do people become what they aim for?",
        "Why people with ADHD either become the greatest or nothing"
    ],
    config = {"max_concurrency" : 5}#A Semaphore, letting only 5 questions at once
)
```
