# LangChain Messages

## What are Messages?

Messages are the **standard way LangChain communicates with LLMs**. Instead of sending plain text, LangChain sends **Message objects** containing:

- **Role** → Who is speaking
- **Content** → Actual text, image, audio, etc.
- **Metadata** → Token usage, IDs, tool calls, response info

> Think of messages as chat bubbles with extra information.

---

## Why use Messages?

Every LLM provider expects a different format.

```
Your Code
    │
LangChain Messages
    │
OpenAI / Gemini / Groq / Claude / ...
```

LangChain converts your messages into the provider's required format, so switching models requires little or no code changes.

---

# Text Prompt

Simply pass a string.

```python
model.invoke("What is AI?")
```

Use when:
- One-time questions
- No conversation history
- Quick testing

---

# Message Prompt

Pass a list of messages instead of a string.

```python
messages = [
    SystemMessage(...),
    HumanMessage(...)
]

model.invoke(messages)
```

Use when you need:
- Roles
- Conversation history
- Better control over responses

---

# Message Types

## 1. SystemMessage

Defines **how the model should behave**.

```python
SystemMessage("You are a Python expert.")
```

Examples:
- You are a coding assistant.
- Answer briefly.
- Explain step by step.

> **Changes the model's behavior, not the user's question.**

---

## 2. HumanMessage

Represents the **user's input**.

```python
HumanMessage("Explain REST APIs")
```

Can contain:
- Text
- Images
- Audio
- Files

Optional metadata:

```python
HumanMessage(
    content="Hello",
    name="alice",
    id="msg_123"
)
```

---

## 3. AIMessage

Represents the **model's response**.

Contains:
- Response content
- Token usage
- Tool calls
- Metadata

```python
response = model.invoke(...)
```

The returned object is an **AIMessage**.

---

## 4. ToolMessage

Represents the **result returned by a tool**.

Flow:

```
Human
   ↓
AI requests tool
   ↓
Tool executes
   ↓
ToolMessage
   ↓
AI generates final answer
```

The `tool_call_id` **must match** the ID generated by the AI.

---

# Metadata

Useful information attached to messages.

Example:

```python
response.usage_metadata
```

Output:

```python
{
    "input_tokens": 53,
    "output_tokens": 258,
    "total_tokens": 311
}
```

---

# Conversation Flow

```
SystemMessage
      ↓
HumanMessage
      ↓
AIMessage
      ↓
HumanMessage
      ↓
AIMessage
```

Conversation history is simply a list of messages passed back to the model.

---

# Quick Revision

| Message | Purpose |
|---------|---------|
| **SystemMessage** | Controls model behavior |
| **HumanMessage** | User input |
| **AIMessage** | Model response |
| **ToolMessage** | Tool execution result |

### Remember

- **String** → Simple prompt
- **List of Messages** → Conversation
- **SystemMessage** = Instructions
- **HumanMessage** = Question
- **AIMessage** = Answer
- **ToolMessage** = Tool output
